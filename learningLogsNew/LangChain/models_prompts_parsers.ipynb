````python
client = OpenAI()
def get_completions(prompt, model='gpt-4o'): 
  messages = [{
    'role': 'user', 
    'content': prompt
    }]
  response = client.chat.completions.create(
    model=model, 
    messages=messages, 
    temperature=0,
  )
  return response.choices[0].message.content
````

````python
from langchain.output_parsers import ResponseSchema, StructuredOutputParser
````

````python
gift_schema = ResponseSchema(
  name="gift", 
  description="Was the item purchased\
  as a gift for someone else? \
  Answer True if yes,\
  False if not or unknown.",
  type = 'bool',
)

delivery_days_schema = ResponseSchema(
  name='delivery_days', 
  description="How many days\
  did it take for the product\
  to arrive? If this \
  information is not found,\
  output -1.",
  type='float',)

price_value_schema = ResponseSchema(
  name="price_value",
  description="Extract any\
  sentences about the value or \
  price, and output them as a \
  comma separated Python list.")
````

````python
response_schemas = [
  gift_schema, 
  delivery_days_schema,
  price_value_schema,
]
````

````python
output_parser = StructuredOutputParser.from_response_schemas(response_schemas=response_schemas)

format_instructions = output_parser.get_format_instructions()

print(format_instructions)
````

````python
review_template_2 = """\
For the following text, extract the following information:

gift: Was the item purchased as a gift for someone else? \
Answer True if yes, False if not or unknown.

delivery_days: How many days did it take for the product\
to arrive? If this information is not found, output -1.

price_value: Extract any sentences about the value or price,\
and output them as a comma separated Python list.

text: {text}

{format_instructions}
"""
prompt = ChatPromptTemplate.from_template(review_template_2)
response = chat.invoke(
  prompt.invoke({
    'text': customer_review,
    'format_instructions': format_instructions
  })
)
print(response.content)
````

### Conversation Buffer Memory

````python
from langchain.chains import ConversationChain
from langchain.memory import ConversationBufferMemory, ConversationBufferWindowMemory
````

````python
llm = ChatOpenAI(
  model='gpt-4o', 
  temperature=0.0
)
# stores conversation in memory 
memory = ConversationBufferMemory()

conversation = ConversationChain(
  llm=llm,
  memory=memory,
  verbose=True,
)
````

````python
conversation.predict(input="Hi my name is Usama")
````

````python
conversation.predict(input='What\'s my name?')
````

#### Sequential Chain

````python
from langchain.chains import SimpleSequentialChain, LLMChain 
````

````python
first_prompt = ChatPromptTemplate.from_template(
  "What is the best name to describe a company that makes {product}? Choose one name."
)

chain_one = LLMChain(llm=llm, prompt=first_prompt)
````

````python
second_prompt = ChatPromptTemplate.from_template(
  'Write a 20-word description for the following company: {company}'
)
chain_two = LLMChain(llm=llm, prompt=second_prompt)
````

````python
overall_simple_chain = SimpleSequentialChain(
  chains = [
    chain_one, 
    chain_two
  ], 
  verbose=True
)
````

````python
overall_simple_chain.invoke(
  {
    'input': 'Queen Size Sheet Set'
  }
)
````

````python
name_prompt = ChatPromptTemplate.from_template(
  "What is the best name to describe a company that makes {product}? Choose one name."
)
brand_name_chain = name_prompt | llm | StrOutputParser()
````

````python
brand_name_schema = ResponseSchema(
  name='brand_name', 
  description='The chosen name of the brand')

description_schema = ResponseSchema(
  name='company_description', 
  description='The description of the company')

response_schemas1 = [
  brand_name_schema, 
  description_schema,
]

output_parser = StructuredOutputParser.from_response_schemas(
  response_schemas=response_schemas1
)

format_instructions1 = output_parser.get_format_instructions()
````

````python
description_prompt = ChatPromptTemplate.from_template(
  'Write a 20-word description for the {company}')
description_chain = description_prompt | llm | StrOutputParser()

format_prompt = ChatPromptTemplate.from_template(
  format_instructions1
)

format_chain = format_prompt | llm 
````

````python
chain = brand_name_chain | description_chain | format_chain
response = chain.invoke({'product': 'ThermaBrew Temperature Mug'})
print(response)
````

````python
from langchain_core.tools import tool

players = [
  {
    "player": "Mohamed Salah",
    "weak_foot": 3,
    "player_of_the_month_awards": 2,
    "apps": 38,
    "minutes_played": 3371,
    "goals": 29,
    "assists": 18
  },
  {
    "player": "Alexander Isak",
    "weak_foot": 4,
    "player_of_the_month_awards": 1,
    "apps": 34,
    "minutes_played": 2756,
    "goals": 23,
    "assists": 6
  },
  {
    "player": "Erling Haaland",
    "weak_foot": 3,
    "player_of_the_month_awards": 1,
    "apps": 31,
    "minutes_played": 2742,
    "goals": 22,
    "assists": 3
  },
  {
    "player": "Bryan Mbeumo",
    "weak_foot": 2,
    "player_of_the_month_awards": 0,
    "apps": 38,
    "minutes_played": 3414,
    "goals": 20,
    "assists": 7
  },
  {
    "player": "Ollie Watkins",
    "weak_foot": 4,
    "player_of_the_month_awards": 0,
    "apps": 38,
    "minutes_played": 2598,
    "goals": 16,
    "assists": 8
  }
]

df = pd.DataFrame(players)

@tool
def retrieve_player_stats(name : str) -> str: 
  """Returns the stats of the specified football players"""
  player_info = df[df['player'] == name]
  return player_info.to_string()
# print(retrieve_player_stats('Ollie Watkins'))
````

````python
agent = create_react_agent(llm, tools=[retrieve_player_stats])

messages = agent.invoke(
  {
    'messages': [
      ('human', 'Create a summary of the player: Ollie Watkins')
    ]
  }
)
print(messages['messages'][-1].content)
````
